Stemming is a rule-based text normalisation technique that reduces words to their root form by removing prefixes or suffixes. The resulting form called a stem, may not be a valid or meaningful word in the language.

Each word is processed independently without considering context
The algorithm checks for common suffixes or prefixes
Predefined heuristic rules are applied to strip these affixes
The remaining part of the word is returned as the stem
No grammatical or semantic validation is performed

Techniques Used
Suffix Stripping: Removes common endings like -ing, -ed, -es
Rule-Based Truncation: Applies fixed linguistic rules
Aggressive Reduction: Shortens words for maximum generalization



```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words=['eating', 'eat','eaten',"running", "studying", "smiling", "communication"]
for word in words:
    print(word +"----->" + stemmer.stem(word))
```

Demerit is that it can change the meaning of word

```python
stemmer.stem('History')
```


Snowball Stemmer is an improved version of Porter’s Stemmer, also known as Porter2. It is faster, more aggressive, and supports multiple languages for stemming tasks.

Improved and faster version of Porter’s Stemmer
Removes suffixes more effectively
Supports multiple languages
Commonly used for multilingual text processing
Produces more consistent stemming results

```python
from nltk.stem import SnowballStemmer

snowballStemmer = SnowballStemmer(language='english')

words_to_stem = ['running', 'jumped', 'happily', 'quickly', 'foxes']

stemmed_words = [snowballStemmer.stem(word) for word in words_to_stem]

print("Original words:", words_to_stem)
print("Stemmed words:", stemmed_words)
```

```python
print(stemmer.stem('happily'))
print(snowballStemmer.stem('happily'))
```

Regexp Stemmer is a flexible stemming algorithm that uses regular expressions (regex) to remove word suffixes based on custom rules.

```python
from nltk.stem import RegexpStemmer

custom_rule = r'ing$'
regexp_stemmer = RegexpStemmer(custom_rule)

word = 'running'
stemmed_word = regexp_stemmer.stem(word)

print(f'Original Word: {word}')
print(f'Stemmed Word: {stemmed_word}')
```

Lancaster Stemmer is a fast and aggressive stemming algorithm that reduces words to very short root forms using iterative rule-based processing.

Faster and more aggressive than many other stemmers
Applies rules repeatedly to reduce words
Can produce very short or distorted stems
Useful for applications requiring fast text processing
May reduce accuracy due to over-stemming

```python
from nltk.stem import LancasterStemmer

stemmer = LancasterStemmer()

words_to_stem = ['running', 'jumped', 'happily', 'quickly', 'foxes']

stemmed_words = [stemmer.stem(word) for word in words_to_stem]

print("Original words:", words_to_stem)
print("Stemmed words:", stemmed_words)
```
