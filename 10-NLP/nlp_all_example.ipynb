One-hot encoding converts categorical text data into binary 0 and 1 values in separate columns so machine learning models can process them

One-Hot Encoding creates a separate column for each category in the dataset


Advantages
Converts categorical data into numerical format for machine learning models
Helps improve model performance by representing categories clearly
Prevents incorrect ordinal relationships between categories

Limitations
Increases the number of columns in the dataset
Can create sparse datasets with many 0 values
May lead to overfitting when categories are too many and data is limited
Spare matrix issue
No semantic meaning is getting captured
Out of Vocabulary

```python
# OHE
from sklearn.preprocessing import OneHotEncoder

# Tokenize the sentence
words = corpus.split()

# OneHotEncoder expects a 2D array
words_2d = [[word] for word in words]

encoder = OneHotEncoder(sparse_output=False)
encoded_data = encoder.fit_transform(words_2d)

print("Words:", words)
print("Encoded data:\n", encoded_data)
print("Categories:", encoder.categories_)
```

Bag of words:

(BoW) model is a simple text-processing technique in Natural Language Processing (NLP) that converts variable-length text into numerical vectors by counting word frequencies while ignoring grammar and word order

Advantages

Simplicity: It is easy to implement and computationally efficient.

Versatility: It can be used for various NLP tasks such as text classification, sentiment analysis and document clustering.

Interpretability: The resulting vectors are interpretable which makes it easy to understand which words are most important in a document.


Limitations

Loss of Context: It ignores word order and context which means it might miss important relationships between words.

Sparsity: When working with large datasets, most word vectors will be sparse (containing mostly zeros) which can lead to inefficiency.

Limited Semantic Understanding: The model doesn’t capture the meaning of words which can be important for some NLP tasks.







```python
from sklearn.feature_extraction.text import CountVectorizer
bog_word= [
    "Hi this is a test for NLP",
    "This is another NLP test",
    "Hi NLP is useful"
]

vectorizer = CountVectorizer()

# Convert text into a word-count matrix
bow_matrix = vectorizer.fit_transform(bog_word)

# Vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

# Convert sparse matrix to a normal NumPy array
# Words are converted to lowercase by default, and words such as "a" may be ignored because CountVectorizer defaults to keeping tokens with at least two characters.
print("\nBag of Words matrix:")
print(bow_matrix.toarray())
```

# TF-IDF

**TF-IDF** stands for **Term Frequency–Inverse Document Frequency**.

It assigns a high score to words that are frequent in one document but uncommon across the whole collection of document

Advantages

Simple and fast: Easy to calculate and efficient for large text collections.

Reduces the importance of common words: Words appearing in many documents receive lower weights.

Highlights important terms: Rare but meaningful words receive higher scores.

Works well with traditional machine-learning models: Commonly used with Naive Bayes, logistic regression, and linear SVM.

Interpretable: You can understand which words contribute to a document’s representation.

Requires no training labels: It is an unsupervised feature-extraction method.

Disadvantages

Ignores word order: "dog bites man" and "man bites dog" may have similar word representations.

Does not understand meaning: "car" and "automobile" are treated as unrelated words.

Produces sparse matrices: Large vocabularies can create many zero values.

Struggles with synonyms: Different words with similar meanings receive separate features.

Sensitive to vocabulary: New or unseen words may be ignored.

Limited context understanding: It does not understand sarcasm, negation, or relationships between words.

Less effective for semantic tasks: Word embeddings or transformer models are usually better for understanding context and meaning.

In short, Bag of Words counts words, while TF-IDF counts words and reduces the weight of words that appear in many documents.


```python
from sklearn.feature_extraction.text import TfidfVectorizer

corpus = [
    "cat sits on mat",
    "cat eats fish",
    "dog eats fish"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(corpus)

# Display vocabulary
print("Vocabulary:")
print(vectorizer.get_feature_names_out())

# Display TF-IDF values
print("\nTF-IDF matrix:")
print(tfidf_matrix.toarray())
import pandas as pd

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print(tfidf_df)
```
