```
import nltk

# sentence to paragraphs
from nltk.tokenize import sent_tokenize

# Download the required tokenizer data
# nltk.download('punkt')
# nltk.download('punkt_tab')

# Define your text data
corpus = "Hello world. This is a test sentence ! test nlp."

# Tokenize into sentences
sentences = sent_tokenize(corpus)
print(sentences)

print(type(sentences))

for sentence in sentences:
    print(sentence)
```

```
from nltk.tokenize import word_tokenize

word_tokenize(corpus)
```

```
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```
# Contraction Splitting: Notice how "Don't" is split into "Do" and "n't", and "you'll" is split into "you" and "'ll".
# Punctuation Isolation: The comma (,) and period (.) are separated into their own distinct tokens.
# Possessives: The trailing possessive "tokenizer's" is split into "tokenizer" and "'s".


from nltk.tokenize import TreebankWordTokenizer

# Instantiate the tokenizer
tokenizer = TreebankWordTokenizer()

# Sample sentence with a contraction and punctuation
text = "Don't blink. or you'll miss the tokenizer's behavior."

# Tokenize the text
tokens = tokenizer.tokenize(text)
print(tokens)
```
