```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good',
]
```

```python
#define the vocabulary size

voc_size=10000
```

```python
# One Hot Encoding

one_hot_rep= [one_hot(words,voc_size)  for words  in sent]
one_hot_rep
```

```python
# Word Embedding

from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences  #need to set sentence  length if 8 and if sentence length is than that add 0 before or after
from tensorflow.keras.models import Sequential
```

```python
import numpy as np

sent_length=8

embedding_docs= pad_sequences(one_hot_rep,maxlen=sent_length)
print(embedding_docs)
```

```python
# feature representation

dim=10
```

```python
model =Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')

```

```python
model.summary
```

```python
model.predict(embedding_docs)
```
